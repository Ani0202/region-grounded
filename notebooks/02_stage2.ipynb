# 02 — Stage 2: Qwen2-VL captioning + SigLIP filter

Captions each Stage-1 crop with Qwen2-VL-2B-Instruct, then drops any
(crop, caption) pair whose SigLIP cosine similarity is below
`cfg.stage2.filter_threshold`. The filtered set is saved as a parquet
that Stage 3 consumes.

```python
%cd /content/region-grounded
import json
from region_grounded import load_config
from region_grounded.stage2_caption import caption_and_filter, save_pairs

cfg = load_config('configs/default.yaml')
records = [json.loads(l) for l in open('outputs/stage1_records.jsonl') if l.strip()]
print('loaded', len(records), 'records from Stage 1')
```

```python
filtered = caption_and_filter(records, cfg.stage2)
kept_regions = sum(len(r['regions']) for r in filtered)
print('kept', kept_regions, 'regions across', len(filtered), 'images')
save_pairs(filtered, cfg.data.pairs_out)
```
